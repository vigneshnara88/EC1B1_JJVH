````python
import matplotlib.pyplot as plt
import numpy as np

# Function to detect outliers using the IQR method
def detect_outliers_iqr(merged_df, columns):
    outliers = {}
    for col in columns:
        Q1 = merged_df[col].quantile(0.25)
        Q3 = merged_df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers[col] = merged_df[(merged_df[col] < lower_bound) | (merged_df[col] > upper_bound)][['date', col]]

    return outliers

# Select only numeric columns for outlier detection
numeric_columns = merged_df.select_dtypes(include=[np.number]).columns

# Detect outliers
outliers_dict = detect_outliers_iqr(merged_df, numeric_columns)

# Plot boxplots for each numeric column to visualize outliers
fig, axes = plt.subplots(nrows=len(numeric_columns)//3 + 1, ncols=3, figsize=(15, 10))
axes = axes.flatten()

for i, col in enumerate(numeric_columns):
    if i < len(axes):  # Ensure we don't exceed available axes
        merged_df.boxplot(column=[col], ax=axes[i])
        axes[i].set_title(f'Boxplot of {col}')

plt.tight_layout()
plt.show()

````

## 🚨 Outlier Detection with Isolation Forest

To ensure data quality and remove anomalies, we used the **Isolation Forest** algorithm — a robust and efficient method for detecting outliers in high-dimensional datasets.

### 🧠 How It Works:
- **Isolation Forest** identifies anomalies by randomly selecting features and splitting values between the min and max.
- Outliers are **isolated faster** than normal observations because they differ significantly from the rest of the data.
- The algorithm assigns:
  - `-1` to outliers
  - `1` to normal points

### 🔧 Steps Taken:
1. **Selected Numeric Columns**: Only numerical features were used for outlier detection.
2. **Standardized the Data**: Applied `StandardScaler` to normalize data for consistent results.
3. **Applied Isolation Forest**:
   - Set `contamination=0.02`, meaning the algorithm flags approximately 2% of the data as outliers.
4. **Flagged Outliers**:
   - Added a new column `outlier_flag` to label each row as `'Normal'` or `'Outlier'`.

### 📁 Output:
- The resulting dataframe allows us to filter and examine outliers separately.
- Detected outliers are stored in `**outliers_df**` for further inspection or removal.


````python
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

# Prepare the dataset for advanced outlier detection
numeric_cols = merged_df.select_dtypes(include=[np.number]).columns

# Standardize the data for better anomaly detection
scaler = StandardScaler()
scaled_data = scaler.fit_transform(merged_df[numeric_cols])

# Apply Isolation Forest (anomaly detection algorithm)
iso_forest = IsolationForest(n_estimators=100, contamination=0.02, random_state=42) #contamination selects for 0.02 (2%) of dataset as outliers
outlier_predictions = iso_forest.fit_predict(scaled_data)

# Mark outliers (-1 means outlier, 1 means normal)
merged_df['outlier_flag'] = outlier_predictions
merged_df['outlier_flag'] = merged_df['outlier_flag'].replace({1: 'Normal', -1: 'Outlier'})

# Extract detected outliers
outliers_df = merged_df[merged_df['outlier_flag'] == 'Outlier']
````

````python
def plot_exchange_rate_graphs(graph_1_df):
    # Ensure that the 'date' column is in datetime format
    graph_1_df['date'] = pd.to_datetime(graph_1_df['date'])
    
    # Create a 2x2 subplot layout
    fig, axes = plt.subplots(2, 2, figsize=(16, 12), sharex=True)
    
    # Flatten axes array for easier indexing
    ax1, ax2, ax3, ax4 = axes.flatten()
    
    # 1. Plot Nominal Exchange Rate Level (top left)
    sns.lineplot(x=graph_1_df['date'], y=graph_1_df['nominal_exchange_rate_japan'], 
                ax=ax1, label='Nominal Exchange Rate', color='blue', linewidth=2)
    ax1.set_title('Nominal Exchange Rate', fontsize=14, fontweight='bold')
    ax1.set_ylabel('Yen per Dollar', fontsize=12)
    ax1.set_xlabel('Date', fontsize=12)  # Added x-axis label
    
    # 2. Plot Monthly Growth in Nominal Exchange Rate (top right)
    sns.lineplot(x=graph_1_df['date'], y=graph_1_df['nominal_ex_rate_growth_japan'], 
                ax=ax2, label='Growth', color='red', linewidth=2)
    ax2.set_title('Nominal Exchange Rate Growth', fontsize=14, fontweight='bold')
    ax2.set_ylabel('Monthly % Change', fontsize=12)
    ax2.set_xlabel('Date', fontsize=12)  # Added x-axis label
    
    # 3. Plot Real Exchange Rate Index (bottom left)
    sns.lineplot(x=graph_1_df['date'], y=graph_1_df['real_exchange_rate_index'], 
                ax=ax3, label='Real Exchange Rate Index', color='green', linewidth=2)
    ax3.set_title('Real Exchange Rate Index', fontsize=14, fontweight='bold')
    ax3.set_xlabel('Date', fontsize=12)
    ax3.set_ylabel('Index (2015=100)', fontsize=12)
    
    # 4. Plot Monthly Growth in Real Exchange Rate (bottom right)
    sns.lineplot(x=graph_1_df['date'], y=graph_1_df['real_ex_rate_growth_japan'], 
                ax=ax4, label='Growth', color='purple', linewidth=2)
    ax4.set_title('Real Exchange Rate Growth', fontsize=14, fontweight='bold')
    ax4.set_xlabel('Date', fontsize=12)
    ax4.set_ylabel('Monthly % Change', fontsize=12)
    
    # Format x-axis ticks for all subplots
    for ax in [ax1, ax2, ax3, ax4]:
        # Create custom ticks for every 5 years
        regular_ticks = [pd.to_datetime(f"{year}-01-01") for year in range(1960, 1995, 5)]
        
        # Add key historical dates
        august_1971 = pd.to_datetime('1971-08-01')
        march_1973 = pd.to_datetime('1973-03-01')
        
        # Combine all ticks and sort them
        all_ticks = regular_ticks + [august_1971, march_1973]
        all_ticks.sort()
        
        # Set the ticks
        ax.set_xticks(all_ticks)
        
        # Create corresponding labels
        tick_labels = []
        for tick in all_ticks:
            if tick.month == 8 and tick.year == 1971:
                tick_labels.append('Aug 1971')
            elif tick.month == 3 and tick.year == 1973:
                tick_labels.append('Mar 1973')
            else:
                tick_labels.append(f'Jan {tick.year}')
        
        # Set the tick labels
        ax.set_xticklabels(tick_labels, rotation=45)
        
        # Add vertical lines at August 1971 and March 1973 with improved visibility
        ax.axvline(x=august_1971, color='black', linestyle='--', linewidth=1.5, alpha=0.7)
        ax.axvline(x=march_1973, color='black', linestyle='--', linewidth=1.5, alpha=0.7)
        
        # Adjust x-axis limits with slight padding
        ax.set_xlim(pd.to_datetime('1959-07-01'), pd.to_datetime('1990-07-01'))
        
        # Add grid for better readability
        ax.grid(True, linestyle='--', alpha=0.7)
    
    # Add annotation for August 1971 to the top left plot (ax1)
    ax1.annotate('Japan Leaving Bretton Woods', 
               xy=(august_1971, 0.7),  # Arrow tip position
               xytext=(pd.to_datetime('1960-01-01'), 0.7),  # Text position
               xycoords=('data', 'axes fraction'),
               textcoords=('data', 'axes fraction'),
               arrowprops=dict(arrowstyle='->', lw=1.5, color='black'),
               fontsize=12, 
               fontweight='bold',
               bbox=dict(boxstyle="round,pad=0.5", fc="lightgrey", ec="black", alpha=0.8))
    
    # Add annotation for March 1973 to the top left plot (ax1)
    ax1.annotate('Floating Exchange Rate System', 
               xy=(march_1973, 0.8),  # Arrow tip position
               xytext=(pd.to_datetime('1960-01-01'), 0.8),  # Text position
               xycoords=('data', 'axes fraction'),
               textcoords=('data', 'axes fraction'),
               arrowprops=dict(arrowstyle='->', lw=1.5, color='black'),
               fontsize=12, 
               fontweight='bold',
               bbox=dict(boxstyle="round,pad=0.5", fc="lightgrey", ec="black", alpha=0.8))
    
    # Add annotation for August 1971 to the bottom left plot (ax3)
    ax3.annotate('Japan Leaving Bretton Woods', 
               xy=(august_1971, 0.2),  # Arrow tip position
               xytext=(pd.to_datetime('1960-01-01'), 0.2),  # Text position
               xycoords=('data', 'axes fraction'),
               textcoords=('data', 'axes fraction'),
               arrowprops=dict(arrowstyle='->', lw=1.5, color='black'),
               fontsize=12, 
               fontweight='bold',
               bbox=dict(boxstyle="round,pad=0.5", fc="lightgrey", ec="black", alpha=0.8))
    
    # Add annotation for March 1973 to the bottom left plot (ax3)
    ax3.annotate('Floating Exchange Rate System', 
               xy=(march_1973, 0.3),  # Arrow tip position
               xytext=(pd.to_datetime('1960-01-01'), 0.3),  # Text position
               xycoords=('data', 'axes fraction'),
               textcoords=('data', 'axes fraction'),
               arrowprops=dict(arrowstyle='->', lw=1.5, color='black'),
               fontsize=12, 
               fontweight='bold',
               bbox=dict(boxstyle="round,pad=0.5", fc="lightgrey", ec="black", alpha=0.8))
    
    # Add a single legend to each subplot with better positioning
    for ax in [ax1, ax2, ax3, ax4]:
        ax.legend(loc='best', framealpha=0.9)
    
    # Add an overall title
    plt.suptitle('Exchange Rate Indicators: Japan (1960-1990)', fontsize=16, fontweight='bold', y=0.98)
    
    # Adjust spacing between subplots
    plt.tight_layout(rect=[0, 0.02, 1, 0.96])  # Adjusted to provide more space at bottom for x-labels
    
    # Create directory if it doesn't exist
    import os
    os.makedirs('./plots', exist_ok=True)
    
    # Save figure before showing it
    plt.savefig('./plots/exchange_rate_graphs.png', dpi=300, bbox_inches='tight')
    
    # Show the plot
    plt.show()
````

````python
def calculate_standard_deviations(df): 
    during_bretton_wood = df[(df['date'] >= '1960-01-01') & (df['date'] <= '1971-06-01')]
    after_bretton_wood = df[(df['date'] >= '1973-03-01')]

    variables = [
        'nominal_ex_rate_growth_japan',  # Monthly growth of nominal exchange rate
        'real_ex_rate_growth_japan',     # Monthly growth of real exchange rate
        'inflation_rate_japan',               # Inflation rate in Japan
        'difference_in_inflation_rates',                # Difference between Japan and US inflation
        'ind_prod_growth_12m_japan'          # 12-month industrial production growth
    ]
     
    results = []
    for var in variables:
        during_std = during_bretton_wood[var].std()
        after_std = after_bretton_wood[var].std()
        ratio = during_std / after_std
        
        results.append({
            'Variable': var,
            'Pre-Bretton Woods SD': during_std,
            'Post-Bretton Woods SD': after_std,
            'Ratio (Post/Pre)': ratio
        })
    
    # Create DataFrame with results
    results_df = pd.DataFrame(results)

    return results_df

standard_deviations = calculate_standard_deviations(merged_df)
````

````python
def create_markdown_table(df):
    # Create better variable names for display
    variable_names = {
        'nominal_ex_rate_growth_japan': 'Nominal Exchange Rate Growth',
        'real_ex_rate_growth_japan': 'Real Exchange Rate Growth',
        'inflation_rate_japan': 'Japan Inflation Rate',
        'difference_in_inflation_rates': 'Japan-US Inflation Difference',
        'ind_prod_growth_12m_japan': 'Industrial Production Growth (12m)'
    }
    
    # Format the values to 4 decimal places
    df_formatted = df.copy()
    df_formatted['Variable'] = df_formatted['Variable'].map(variable_names)
    df_formatted['Pre-Bretton Woods SD'] = df_formatted['Pre-Bretton Woods SD'].map('{:.4f}'.format)
    df_formatted['Post-Bretton Woods SD'] = df_formatted['Post-Bretton Woods SD'].map('{:.4f}'.format)
    df_formatted['Ratio (Post/Pre)'] = df_formatted['Ratio (Post/Pre)'].map('{:.4f}'.format)
    
    # Create markdown table header
    markdown_table = "### Standard Deviations Before and After Bretton Woods\n\n"
    markdown_table += "| Variable | Pre-Bretton Woods SD | Post-Bretton Woods SD | Ratio (Pre/Post) |\n"
    markdown_table += "|:---------|:--------------------:|:---------------------:|:----------------:|\n"
    
    # Add each row
    for _, row in df_formatted.iterrows():
        markdown_table += f"| {row['Variable']} | {row['Pre-Bretton Woods SD']} | {row['Post-Bretton Woods SD']} | {row['Ratio (Post/Pre)']} |\n"
    
    return markdown_table

# Generate the markdown table
markdown_table = create_markdown_table(standard_deviations)
print(markdown_table)
````

## 📉 Regression Analysis
We perform **two regression models**:
1. **Inflation Difference Volatility vs. Real Exchange Rate Volatility** – To understand if real exchange rate volatility influences inflation differences.
2. **Industrial Production Growth Volatility vs. Real Exchange Rate Volatility** – To examine the impact of exchange rate volatility on industrial production growth.

### **Regression 1: Inflation Difference Volatility**
- **Independent Variable**: Real Exchange Rate Growth (Japan) Volatility
- **Dependent Variable**: Japan-US Inflation Difference Volatility

The regression summary shows how exchange rate volatility affects inflation difference volatility.

### **Regression 2: Industrial Production Growth Volatility**
- **Independent Variable**: Real Exchange Rate Growth (Japan) Volatility
- **Dependent Variable**: Industrial Production Growth Volatility

This regression analyzes whether fluctuations in real exchange rates contribute to industrial production volatility.

````python
def create_six_month_intervals(df):
    # Make a copy of the dataframe to avoid modifying the original
    df_copy = df.copy()
    
    # Ensure date column is datetime
    df_copy['date'] = pd.to_datetime(df_copy['date'])
    
    # Extract year and month
    df_copy['year'] = df_copy['date'].dt.year
    df_copy['month'] = df_copy['date'].dt.month
    
    # Create semester indicator (1 for Jan-Jun, 2 for Jul-Dec)
    df_copy['semester'] = ((df_copy['month'] - 1) // 6) + 1
    
    # Create period label in format "YYYY/S" where S is 1 or 2
    df_copy['period'] = df_copy['year'].astype(str) + '/' + df_copy['semester'].astype(str)
    
    return df_copy

# Apply the function to create intervals
merged_df_with_periods = create_six_month_intervals(merged_df)
display(merged_df_with_periods)

# If you want to calculate standard deviations by period instead of means:
period_std = merged_df_with_periods.groupby('period').agg({
    'real_ex_rate_growth_japan': 'std',
    'difference_in_inflation_rates': 'std',
    'ind_prod_growth_12m_japan': 'std',
})

display(period_std)
# To remove the last row which is NaN 
period_std = period_std[:-1]
````

````python
import statsmodels.api as sm

X_1 = period_std["real_ex_rate_growth_japan"]  # Independent variable
X_1 = sm.add_constant(X_1)  # Adds a constant for regression

y1 = period_std["difference_in_inflation_rates"]  # Dependent variable 1

# Run regression
model1 = sm.OLS(y1, X_1).fit()

# Print detailed regression summary
print("Regression 1: Inflation Difference Volatility")
print(model1.summary())
````
